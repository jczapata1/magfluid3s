**Automatization**

````python
from libs.magfluid3s_auto import MagFluid3SAuto
````

````python
auto = MagFluid3SAuto(simulation='MvsH',
                      solver='llg',
                      input_file='./examples/MvsH/Input.in',
                      output_folder='./examples/Auto_MvsH/',
                      properties={'RM':6e-9, 'T0':300},
                      n=5) 
````

````python
auto.run()          # Execution    
auto.make_summary() # Data Processing
auto.plot_summary() # Visualization
````

**Quick Guide**

`simulation:`

* Microstates: magnetic and structural response under constant temperature and magnetic field.  
* MvsH: field-dependent magnetic response under constant temperature. Hysteresis loop.
* MvsT: temperature-dependent magnetic response under constant magnetic field. ZFC and FC protocols.  

`solver:`

* llg: Landau–Lifshitz–Gilbert equation $\left( \frac{d\hat{m}}{dt} \right)$.  
* llg-t: Landau–Lifshitz–Gilbert and Torque equations $\left( \frac{d\hat{m}}{dt}, \frac{d\hat{n}}{dt} \right)$.

`input_file:`  
Where is the input file?  

`project_folder:`    
Where is the project folder?  

`properties:`  
```
{'prop_sym':prop__val, ...}
```
* prop_sym: property symbol.
* prop_val: property value.
  
`n:`   
How many experiments?
